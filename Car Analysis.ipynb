```python
import matplotlib.pyplot as plt

car['Origin'].value_counts().plot(kind='bar')

plt.title('Number of Cars by Origin')
plt.xlabel('Origin')
plt.ylabel('Number of Cars')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(car['Horsepower'], car['MPG_City'], alpha=0.7)

plt.title('Horsepower vs City Mileage')
plt.xlabel('Horsepower')
plt.ylabel('City MPG')
plt.tight_layout()
plt.show()
```
